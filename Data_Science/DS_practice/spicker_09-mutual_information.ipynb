```python
# extern

import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

from matplotlib import pyplot as plt

# --- Zufällige Datenmatrix ---
np.random.seed(420)
X, T = make_blobs(n_samples=150, centers=5, cluster_std=1, random_state=40)

# --- Clusterung -> y ---
K = 8
kmeans = KMeans(n_clusters=K, n_init="auto", random_state=42)
C = kmeans.fit_predict(X)

# --- Visualisierung ---
# plt.scatter(*X.T, c=y, label="y")
# plt.scatter(*X.T, marker="x", c=y_, label="predicted")
# plt.legend()
# plt.show()

# --- Wahrscheinlichkeit, dass ein Punkt aus Cluster i zur Partition j aus T gehört ---
p = np.array([[np.logical_and(C==i, T==j).sum() for j in np.unique(T)] for i in np.unique(C)]) / X.shape[0]

# --- Wahrscheinlichkeit für ein Cluster C_i ---
p_C = np.array([(C==i).sum() / X.shape[0] for i in np.unique(C)]) / X.shape[0]

# --- Wahrscheinlichkeit für ein Cluster T_j ---
p_T = np.array([(T==j).sum() / X.shape[0] for j in np.unique(T)]) / X.shape[0]

# --- Mutual information -> I_CT ---
I_CT = np.sum([[p[i,j] * np.log2(p[i,j] / (p_C[i] * p_T[j])) if p_C[i] * p_T[j] != 0 and p[i,j] != 0 else 0 for j in np.unique(T)] for i in np.unique(C)])
```
