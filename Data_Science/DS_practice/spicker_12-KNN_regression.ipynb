```python
import numpy as np
from sklearn.model_selection import train_test_split

from matplotlib import pyplot as plt

# --- Datenmatrix ---
X = np.array(range(150)) / 5
y = np.sin(X)

# --- Aufteiling in train und test Daten ---
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42)
idx = X_train.argsort()
X_train = X_train[idx]
y_train = y_train[idx]

# --- Verwende Mittelwert der labels der Nachbarn --- 
K = 5
y_pred = np.array([y_train[np.argsort([np.linalg.norm(x - x_) for x_ in X_train])][:K].mean(axis=0) for x in X_test])

# --- Visualisierung ---
# plt.plot(X_train, y_train, marker="x", alpha=.25, label="train")
# plt.scatter(X_test, y_test, label="test")
# plt.scatter(X_test, y_pred, label="pred")
# plt.legend()
# plt.show()
```
