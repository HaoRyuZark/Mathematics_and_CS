```python
import numpy as np
from sklearn.decomposition import PCA

from matplotlib import pyplot as plt

# --- Zufällige Datenmatrix ---
np.random.seed(42)
X = np.random.rand(42, 10).T

# --- zscore ---
Z = (X - X.mean(axis=0)) / X.std(axis=0)
# Alternativ: Nicht normierte Daten
# Z = X

# --- Kovarianzmatrix ---
cov_mat = np.cov(Z.T)

# --- Eigenwertzerlegung ---
eig_vals, eig_vecs = np.linalg.eigh(cov_mat)

# --- Eigenvektoren nach Eigenwerten sortieren ---
idx = np.argsort(eig_vals)[::-1]
eig_vals = eig_vals[idx]
eig_vecs = eig_vecs[:,idx]

# --- Propotion of Variance Explained -> PVE ---
PVE = eig_vals / eig_vals.sum()

# --- Visualisierung ---
# plt.plot(range(1, PVE.shape[0]+1), PVE, marker="x")
# plt.xlabel("PCA-Komponente")
# plt.ylabel("PVE")
# plt.show()

# plt.plot(range(1, PVE.shape[0]+1), PVE.cumsum(), marker="x")
# plt.xlabel("PCA-Komponente")
# plt.ylabel("kumulative PVE")
# plt.show()

# --- Alle PCA-Komponenten verwenden, die mehr als 10% der Varianz ausmachen ---
idx = PVE > 0.1
# Alternativ: Verwende 2-PCA Komponenten zur Visualisierung
# idx = range(2)
eig_vals = eig_vals[idx]
eig_vecs = eig_vecs[:,idx]

# --- Projizieren der Daten auf die PCA-Komponenten -> C ---
C = np.array([[z @ eig_vec for eig_vec in eig_vecs.T] for z in Z])
# Alternativ: Wenn auf X gearbeitez wird: for z in X - X.mean(axis=0)
# C = np.array([[z @ eig_vec for eig_vec in eig_vecs.T] for z in Z - Z.mean(axis=0)])

# --- Vergleichswert ---
# Obacht: Vorzeichen der Eigenvektoren nicht eindeutig
# pca = PCA(n_components=eig_vecs.shape[1])
# C_ = pca.fit_transform(Z)
# print(pca.components_)          # eig_vecs
# print(pca.explained_variance_)  # eig_vals

# --- Visualisierung ---
# assert C.shape[1] == 2, f"{C.shape=}"
# plt.scatter(*C.T, label="spicker")
# plt.scatter(*C_.T, marker="x", label="sklearn")
# plt.legend()
# plt.xlabel("1. PCA-Komponente")
# plt.ylabel("2. PCA-Komponente")
# plt.show()

# --- Rekonstruktion von Z -> G ~ X ---
G = np.vstack([np.sum([k * eig_vec for k, eig_vec in zip(c, eig_vecs.T)], axis=0) for c in C])
# Alternativ: Wenn auf X gearbeitez wird: + X.mean(axis=0)
# G = np.vstack([np.sum([k * eig_vec for k, eig_vec in zip(c, eig_vecs.T)], axis=0) for c in C]) + Z.mean(axis=0)
```
