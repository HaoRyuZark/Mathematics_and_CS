```python
# extern

import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

from matplotlib import pyplot as plt

# --- Zufällige Datenmatrix ---
np.random.seed(420)
X, T = make_blobs(n_samples=150, centers=5, cluster_std=1, random_state=40)

# --- Clusterung -> y ---
K = 8
kmeans = KMeans(n_clusters=K, n_init="auto", random_state=42)
C = kmeans.fit_predict(X)

# --- Visualisierung ---
# plt.scatter(*X.T, c=T, label="y")
# plt.scatter(*X.T, marker="x", c=C, label="predicted")
# plt.legend()
# plt.show()

# --- purity nach cluster ---
purity = [1/X[C==i].shape[0] * max([np.logical_and(C==i, T==j).sum() for j in np.unique(T)]) for i in np.unique(C)]

# --- purity ---
purity = 1/X.shape[0] * sum([max([np.logical_and(C==i, T==j).sum() for j in np.unique(T)]) for i in np.unique(C)])
```
