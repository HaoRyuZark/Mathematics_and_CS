```python
xi = Matrix([-3, 0, 4])
p = lambda x : 3*x - 2*x**2
yi = Matrix([p(x) for x in xi])

Math(r'x_i = {}, \quad y_i = {}'.format(list(xi), list(yi)))
```

# Normalgleichung

Für das Ausgleichspolynom muss 

\begin{align*}
\sum_{i=0}^2 (p(x_i)-y_i)^2
= \sum_{i=0}^2 (a x_i + b x_i^2-y_i)^2
= 
\big\|
\underbrace{
\begin{pmatrix}
x_0 & x_0^2 \\
x_1 & x_1^2 \\
x_2 & x_2^2 \\
\end{pmatrix}
}_A
\underbrace{
\begin{pmatrix}
a \\ b
\end{pmatrix}
}_v
-
\underbrace{
\begin{pmatrix}
y_0 \\ y_1 \\ y_2 
\end{pmatrix}
}_y
\big\|_2^2
= \big\| Av-y \big\|_2^2
\end{align*}
in $v$ minimiert werden. Da $A$ vollen Rang hat, ist der Minimierer die eindeutige Lösung der Normalgleichung

\begin{align*}
A^TA v= A^Ty.
\end{align*}

```python
xi0 = Matrix([x**1 for x in xi])
xi2 = Matrix([x**2 for x in xi])

A = xi0.row_join(xi2)
display(Math('A=' + platex(A)))

ATA = A.T*A
display(Math('A^T A=' + platex(ATA)))


ATy = A.T*yi
display(Math('A^T y=' + platex(ATy)))

xx = ATA.solve(ATy)


ATA = A.T*A
display(Math('v = (a,b)=' + platex(xx.T)))

x = symbols('x')

p = (xx.T * Matrix([x,x**2]))[0]

display(Math('p(x)=' + latex(p)))

#display(A.solve_least_squares(yi))
```

# QR Zerlegung nach Givens

$A\in \mathbb{R}^{m \times n},\ m \ge n,\ \mathrm{rang} (A) = n,\ b \in \mathbb{R}^n$ gegeben,
bestimme eindeutiges $v$ mit $\| Av - y \|_2$ minimal durch

- Elimination aller $m$ Spalten von $A$ mit einer orthonormalen Matrix $Q$
  
\begin{align*}
  A \rightarrow QA = \begin{pmatrix} R \\ 0 \end{pmatrix}, \qquad y
\rightarrow Qy = \begin{pmatrix} c \\d \end{pmatrix}
\end{align*}

- Lösen von $Rv = c$ durch rückwärts einsetzen

```python
def givens(AA, yy):
    A = AA.copy()
    y = yy.copy()
    m,n = A.shape
    QQ = eye(m)
    for j in range(n):
        for i in range(j+1,m):
            no = sqrt(A[j,j]**2 + A[i,j]**2)
            c = A[j,j] / no
            s = A[i,j] / no
            Q = eye(m)
            Q[j,j], Q[j,i] =  c, s
            Q[i,j], Q[i,i] = -s, c
            out = r'Q_{{{0}{1}}} = {2},\quad (QA)_{{{0}{1}}} = {2}{3} = {4},\quad (Qy)_{{{0}{1}}} = {2}{5} = {6}'.format(i+1,j+1,platex(Q),platex(A),platex(Q*A), platex(y), platex(Q*y))
            display(Math(out))
            
            A = Q*A
            y = Q*y
            QQ = QQ*Q.T
    return QQ,A,y
```

```python
Q, QA, Qy = givens(A, yi)
```

Damit ist

```python
display(Math(r'QA = {}, \quad Qy = {}'.format(platex(QA), platex(Qy))))
```

bzw.

```python
R = QA[:2,:2]
c = Qy[:2, 0]

display(Math(r'R = {}, \quad c = {}'.format(platex(R), platex(c))))
```

und somit

```python
v = R.inv() * c

display(Math(r'v = R^{{-1}}c = {}'.format(platex(v))))
```
