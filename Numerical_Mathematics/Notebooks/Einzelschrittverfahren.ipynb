```python
from sympy import *

from IPython.display import display, Math, Latex
from sympy.interactive import printing
printing.init_printing(use_latex='mathjax')
platex = lambda A: latex(A,mat_str='pmatrix',mat_delim='')

import pylab as pl
%matplotlib inline
```

# Aufgabenstellung

Wenden Sie auf das lineare Gleichungssystem $Ax = b$,

```python
A = Matrix([[4, 2, 0], [2, 4, 2], [0, 2, 4]])
b = Matrix([2, 0, 2])
Math('A=' + platex(A) + r',\qquad b=' + platex(b))
```

das Einzelschrittverfahren an.

* Berechnen Sie für $x_0 = (0, 0, 0)^T$ die Iterierten $x_1 ,\ldots, x_4$. 

* Bestimmen Sie die Iterationsmatrix $B_{GS}$ sowie $\|B_{GS}\|_{\infty}$. Konvergiert das Verfahren? Fällt der Fehler $\|e_k\|_{\infty}$ monoton?

# Lösung

## Iteration

```python
from numpy import triu, tril, diag

D = Matrix(diag(diag(A)))
E = - Matrix(tril(A, -1))
F = - Matrix(triu(A,  1))

B = (D - E).inv() * F
d = (D - E).inv() * b

x0 = 0 * b
nit = 4

x = x0
xk  = 'x_0 =' + platex(x)
xkf = 'x_0 =' + platex(x.evalf(6)) 
for it in range(nit):
    x = B * x + d
    xk  += r',\quad x_' + str(it+1) + '=' + platex(x)
    xkf += r',\quad x_' + str(it+1) + r'\approx' + platex(x.evalf(6)) 

display(Math(xk))
```

## Iterationsmatrix und Konvergenz.

Die Iterationsmatrix ist gegeben durch

```python
Math(r'B = (D-E)^{-1}F =' + platex((D-E).inv()) + platex(F) + '=' + platex(B))
```

Damit ist

```python
def norm_inf(A):
    e = ones(A.shape[1])
    return max(A.applyfunc(abs)*e)

Math(r'\|B\|_\infty =' + latex(norm_inf(B)) )
```

Wegen $\rho(B) \leq \|B\|_\infty < 1$ konvergiert das Verfahren. 
Da $\|e_{k+1}\|_\infty \leq \|B\|_\infty \|e_{k}\|_\infty$ und $\|B\|_\infty < 1$ ist, 
ist die Konvergenz in der Unendlich-Norm monoton.
