```python
x = symbols('x')
f = Lambda(x, x / (1 + x))

a = 0
b = 4

Iex = Integral(f(x), (x,a,b))

Math(latex(Iex))
```

Die $P_{ij}$ des Extrapolationstableaus sind durch
$$
P_{ij}=P_{ij-1}+\frac{P_{ij-1}-P_{i-1j-1}}{(\frac{h_{i-j}}{h_i})^2-1},\qquad 1\leq j \leq i, % \leq m,
$$
gegeben. Dabei ist $P_{i0}$ jeweils das Ergebnis der summierten Trapezregel zur Trapezbreite
$h_i$.

## Summierte Trapezregel

Die summierte Trapez-Regel für $n$ Trapeze ist definiert durch
$$
T_n = \frac{h}{2} \big(f(a) + 2f(a+h) + \ldots + 2f(a+(n-1)h) + f(a+nh) \big),
\qquad
h = \frac{b-a}{n}
$$
Für $n=1,2,4$ erhalten wir

```python
def tsum(f, a, b, n):
    h = (b - a) / Rational(n)
    
    w = f(a)
    for i in range(1, n):
        w += 2 * f(a + i*h)
    w += f(b)
    
    return w*h/Rational(2)

st = 7
for n in [1, 2, 4]:
    Tn = tsum(f, a, b, n)
    display( Math(r'T_' + str(n) + ' = '+ latex(Tn) + ' \\approx ' + latex(Tn.evalf(st)) ) )
```

## Romberg

Sei $T$ die zu $q$ Schrittweiten $h_0 > h_1 > \ldots >h_{q-1}$
extrapolierte äquidistante summierte Trapez-Regel. Dann ist $T =
P_{q-1 q-1}$ mit
\begin{align*}
P_{i0} &= T_{m_i} && i = 0, \ldots , q-1\\[1ex]
P_{ij} &= P_{ij-1} + \frac{P_{ij-1} -
P_{i-1j-1}}{(\frac{h_{i-j}}{h_i})^2 - 1} && 1 \le j \le i \le q-1.
\end{align*}
$T$ approximiert $\int\limits_a^b f(t) dt$ für $f \in C^{2q-1}
[a,b]$ mit Ordnung $2q$.

```python
# Extrapolationstableau
def extra(hmi, Tmi):
    q = len(hmi)
    P = zeros(q)
    
    P[:,0] = Matrix(Tmi)
    for i in range(1,q):
        for j in range(1,i+1):
            P[i, j] = P[i, j-1] + (P[i, j-1] - P[i-1, j-1]) / ((hmi[i-j] / hmi[i])**2 - 1)
    
    return P

# Romberg
mi  = [1, 2, 4]
hmi  = [(b - a) / Rational(m)  for m in mi]
Tmi = [tsum(f, a, b, m)  for m in mi]

P = extra(hmi, Tmi)

display(Math(r'm_i = ' + latex(mi)))
display(Math(r'h_i = ' + latex(hmi)))
display(Math(r'T_{m_i} = ' + latex(Tmi) + r'\approx' + latex(Matrix(Tmi).T.evalf(7))))
display(Math(r'P = ' + latex(P) + r'\approx' + latex(P.evalf(7))))

T = P[-1, -1]

Math('Romberg = '+ latex(T.simplify()) + r' \approx ' + latex(T.evalf(7)) )
```
