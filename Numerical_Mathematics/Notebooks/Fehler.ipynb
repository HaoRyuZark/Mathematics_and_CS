# Aufgabenstellung

Bei Parallelschaltung zweier Widerstände $x$ und $y$ berechnet sich der
Gesamtwiderstand $f$ gemäß
$$
\frac{1}{f(x,y)}=\frac{1}{x}+\frac{1}{y}.
$$
Die beiden Widerstände $x=5$, $y=20$ seien mit einer relativen Genauigkeit von jeweils $20\%$ gegeben.

-   Berechnen Sie den maximalen absoluten Fehler.


-   Schätzen Sie den absoluten Fehler von $f$ mit Hilfe der
    Fehlerfortpflanzungsformel
    $$
    |f(x',y')-f(x,y)|\approx |\partial_xf(x,y)|\cdot|x'-x| +
    |\partial_yf(x,y)|\cdot|y'-y|
    $$
    ab.

# Lösung

Die beiden Widerstände $x=5$, $y=20$ seien mit einer relativen Genauigkeit von jeweils $20\%$ gegeben,
d.h. durch die Messfehler können Werte im Bereich

```python
x =  5
y = 20
er = Rational(20, 100)

dx = x * er
dy = y * er

xmin, xmax = x - dx, x + dx
ymin, ymax = y - dy, y + dy

xint = [xmin, xmax]
yint = [ymin, ymax]

Math("x'\in [{},{}], \qquad y'\in [{},{}]".format(*xint, *yint))
```

gemessen werden.

## Maximaler absoluter Fehler

Der maximale absolute Fehler ist die größte Abweichung vom exakten Wert $f(x,y)$,
die durch die Messfehler verursacht werden kann, also
$$
e_{a,\max} = \max_{\substack{ x'\in [x_{\min}, x_{\max}] \\ y\in [y_{\min}, y_{\max}]}}
|f(x',y') - f(x, y)|.
$$
Wir müssen also
$$
g(x',y') = f(x',y') - f(x,y)
$$
betragsmäßig maximieren. Dazu betrachten wir die Minima und Maxima von $g$. Mit

```python
xs,ys = symbols("x',y'")
f = Lambda((xs,ys), 1/(1/xs +1/ys))

display(Math("f(x',y') = " + latex(f(xs,ys))))

g = Lambda((xs,ys), f(xs,ys) - f(x,y))

Math("g(x',y') =" + latex(g(xs,ys)))
```

und

```python
Math(  r"\partial_{x'} g(x',y') = " + latex(g(xs,ys).diff(xs).simplify())  + r',\quad'
     + r"\partial_{y'} g(x',y') = " + latex(g(xs,ys).diff(ys).simplify()))
```

folgt
$$
\partial_{x'} g(x',y') > 0,
\qquad
\partial_{y'} g(x',y') > 0
$$ 
für alle zulässigen $(x',y')$. Damit ist $g$ streng monoton wachsend in beiden Variablen und somit

```python
gmin = g(xmin,ymin)

xyint = "{\substack{ x'\in" + str(xint) + r"\\ y'\in" + str(yint) + r"}}"
display(Math(r"g_{\min} = \min_" + xyint + "g(x',y') = g" + latex((xmin,ymin)) + "=" + latex(gmin)))
                         
gmax = g(xmax,ymax)
display(Math(r"g_{\max} = \max_" + xyint + "g(x',y') = g" + latex((xmax,ymax)) + "=" + latex(gmax)))
```

Damit folgt schließlich

```python
eamax = max([abs(gmin), abs(gmax)])
Math(r'e_{a,\max} = \max(|g_{\min}|,|g_{\max}|) =' + latex(eamax))
```

## Fehlerfortpflanzungsformel

Es ist
$$
|f(x',y')-f(x,y)|\approx |\partial_xf(x,y)|\cdot|x'-x| +
|\partial_yf(x,y)|\cdot|y'-y|
$$
mit

```python
Math('(x,y) = ' + latex((x,y)) 
     + r",\quad |x'-x| \leq" + str(dx)
     + r",\quad |y'-y| \leq" + str(dy))
```

Wegen

```python
xs,ys = symbols("x,y")

fx = Lambda((xs,ys), diff(f(xs,ys),xs).simplify())
fy = Lambda((xs,ys), diff(f(xs,ys),ys).simplify())

Math(r"\partial_{{x}} f({},{}) = ".format(x,y) + latex(fx(x,y)) + r",\quad" + 
     r"\partial_{{y}} f({},{}) = ".format(x,y) + latex(fy(x,y)))
```

folgt

```python
eaf = abs(fx(x,y)) * dx  + abs(fy(x,y)) * dy

Math(r"|\partial_xf(x,y)|\cdot|x'-x| + |\partial_yf(x,y)|\cdot|y'-y| = " + latex(eaf))
```
