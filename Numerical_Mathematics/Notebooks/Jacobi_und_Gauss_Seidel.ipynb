```python
from sympy import *

from IPython.display import display, Math, Latex
from sympy.interactive import printing
printing.init_printing(use_latex='mathjax')
platex = lambda A: latex(A,mat_str='pmatrix',mat_delim='')

import pylab as pl
%matplotlib inline
```

# Aufgabenstellung

Betrachten Sie das lineare Gleichungssytem $Ax=b$ mit

```python
A = Matrix([[2, 3], [1, 2]])
b = Matrix([203, 102])
Math('A=' + platex(A) + r',\qquad b=' + platex(b))
```

* Berechnen Sie 
die Iterierten $x_1,\ldots,x_5$ für das Jacobi- sowie das Gauß-Seidel-Verfahren
zum Startwert $x_0=(0,0)^T$.

* Bestimmen Sie die Iterationsmatrizen $B_J$ uns $B_{GS}$ der beiden Verfahren sowie
deren Spektralradien.

# Lösung

## Iteration

```python
D = Matrix(pl.diag(pl.diag(A)))
E = - Matrix(pl.tril(A, -1))
F = - Matrix(pl.triu(A,  1))

Bj = D.inv() * (E + F)
dj = D.inv() * b

Bgs = (D - E).inv() * F
dgs = (D - E).inv() * b

x0 = zeros(2,1)

def iter(x0, B, d, nit = 5):
    x = x0.copy()
    xk  = 'x_0 =' + platex(x)
    xkf = 'x_0 =' + platex(x.evalf(6)) 
    for it in range(nit):
        x = B * x + d
        xk  += r',\quad x_' + str(it+1) + '=' + platex(x)
        xkf += r',\quad x_' + str(it+1) + r'\approx' + platex(x.evalf(6))
    
    return (xk, xkf)
```

Für Jacobi erhält man

```python
xj, xjf = iter(x0, Bj, dj)

display(Math(xj))
display(Math(xjf))
```

und für Gauß-Seidel

```python
xgs, xgsf = iter(x0, Bgs, dgs)

display(Math(xgs))
display(Math(xgsf))
```

## Iterationsmatrizen und Spektralradien

Für Jacobi ist die Iterationsmatrix gegeben durch

```python
Math(r'B_J = D^{-1}(E+F) =' + platex(D.inv()) + platex(E+F) + '=' + platex(Bj))
```

Die Eigenwerte sind

```python
ev = list(Bj.eigenvals().keys())

out = ''
for i,e in enumerate(ev):
    out += r'\lambda_' + str(i+1) + '=' + latex(e) + r', \quad'

Math(out)
```

der Spektralradius ist dann

```python
rho = max(list(map(abs,ev)))
Math(r'\rho = \max_i |\lambda_i| = ' + latex(rho) + r'\approx' +  latex(rho.evalf(6)) )
```

Analog erhalten wir für Gauß-Seidel

```python
Math(r'B_{GS} = (D-E)^{-1}F =' + platex((D-E).inv()) + platex(F) + '=' + platex(Bgs))
```

mit Eigenwerten

```python
ev = list(Bgs.eigenvals().keys())

out = ''
for i,e in enumerate(ev):
    out += r'\lambda_' + str(i+1) + '=' + latex(e) + r', \quad'

Math(out)
```

und Spektralradius

```python
rho = max(list(map(abs,ev)))
Math(r'\rho = \max_i |\lambda_i| = ' + latex(rho) + r'\approx' +  latex(rho.evalf(6)) )
```
