```python
import sympy as sy
import numpy as np
```

```python
A = sy.Matrix([[1,1,1],[-1,1,1],[-1,-1,1]], dtype=float)
A
```

# Exakte Eigenwerte

```python
ew = A.eigenvals()
ew = sy.Matrix(list(ew.keys()))
ew
```

```python
ew.evalf()
```

# QR Iteration (ohne Shift)

```python
for k in range(10):
    Q,R = np.linalg.qr(A)
    A = R.dot(Q)
    display(A)
```

Die Grenzmatrix ist Block-diagonal.

Der erste Block ist

```python
A1 = A[:2,:2]
A1
```

mit Eigenwerten

```python
np.linalg.eigvals(A1)
```

Der zweite Block ist

```python
A2 = A[2:,2:]
A2
```

mit Eigenwert

```python
np.linalg.eigvals(A2)
```
