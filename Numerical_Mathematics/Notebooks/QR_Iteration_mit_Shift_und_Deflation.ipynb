```python
#from sympy import *

from IPython.display import display, Math, Latex, Markdown
from sympy import Matrix, latex
from sympy.interactive import printing
printing.init_printing(use_latex='mathjax')
platex = lambda A: latex(A,mat_str='pmatrix',mat_delim='')

from pylab import *
from numpy.linalg import qr, eigvalsh

%matplotlib inline
```

# Aufgabenstellung

Approximieren Sie die Eigenwerte der Matrix

```python
A = array([[2,1,0,0], [1,2,1,0], [0,1,2,1], [0,0,1,2,]], dtype = float)
display(Math('A = ' + platex(Matrix(A))))
```

mit Hilfe der QR-Iteration mit Shift und Deflation.

# Lösung

```python
ndigi = 4

Ak = A.copy()
Q = eye(*A.shape)

n = Ak.shape[0]

eps = 0.0001

ndigi = 4

while n > 1:
    display('----')
    display(Math(r'n = {}'.format(n)))
    display('----')
    display(Math('A_k = ' + platex(Matrix(Ak[:n, :n]).evalf(ndigi))))
    while abs(Ak[n-1, n-2]) > eps:
        Bk = Ak[n-2:n, n-2:n]
        display(Math('B_k = ' + platex(Matrix(Bk).evalf(ndigi))))
        gamma = eigvalsh(Bk)
        display(Math(r'\gamma_i \in'+ latex(set(Matrix(gamma).evalf(ndigi)))))
        muk = gamma[ abs(gamma - Ak[n-1,n-1]).argmin() ]
        display(Math(r'\mu_k ='+ latex(round(muk, ndigi))))
        
        Qk, Rk = qr(Ak[:n,:n] - muk * eye(n))
        Ak[:n,:n] = Rk.dot(Qk) + muk * eye(n)
        
        Q[:,:n] = Q[:,:n].dot(Qk)
        
        display(Math('A_k = ' + platex(Matrix(Ak[:n, :n]).evalf(ndigi))))
    n -= 1
```

Als Endergebnis erhalten wir folgende näherungsweise diagonale Matrix, die die Approximationen
der Eigenwerte enthält

```python
display(Math(platex(Matrix(Ak).evalf(ndigi))))
```
