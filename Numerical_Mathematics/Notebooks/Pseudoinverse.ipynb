# Pseudoinverse

Man kann $A^+$ auch dadurch berechnen, dass man für beliebiges $b$ alle $x$, für die
$\|Ax-b\|_2^2$
sein Minimum annimmt mit Hilfe der Normalgleichung bestimmt und dann
unter allen Lösungen der Normalgleichung die mit der kleinsten $\|\cdot\|_2$-Norm sucht.

$\|Ax-b\|_2^2$ wird minimal, falls $A^TAx = A^Tb$ ist, d.h. das Normalgleichungssystem muss (z.B. mit Gauß-Elimination) für beliebiges $b$ gelöst werden.

```python
ATA = A.T*A
display(Math('A^T A=' + latex(ATA)))

class DeferredVectorN(DeferredVector):
   def __getitem__(self, i):
        if i == -0:
            i = 0
        if i < 0:
            raise IndexError('DeferredVector index out of range')
        component_name = r'%s_%d' % (self.name, i)
        return Symbol(component_name)

b = DeferredVectorN('b')
b = Matrix([b[i] for i in range(1, 1 + A.shape[0])])

ATb = A.T*b
display(Math('A^T b=' + latex(ATb)))
```

$A^TA$ hat keinen vollen Rang, somit hat das Normalgleichungssystem mehr als eine Lösung $z$

```python
z = linsolve((ATA, ATb))
z = Matrix(z.args[0])
Math(r'z = {}'.format(platex(z)))
```

Wir müssen jetzt den freien Parameter $\tau_0$ so bestimmen,
dass $\|z\|_2^2$ minimal wird:

```python
z22 = z.dot(z).simplify()
Math(r'f(\tau_0) = \|z\|_2^2 = {}'.format(latex(z22)))
```

```python
z22s = z22.diff('tau0')
Math(r"f'(\tau_0) = {}".format(latex(z22s)))
```

```python
t0 = solve(z22s, 'tau0')[0]
Math(r'\tau_0 = {}'.format(latex(t0)))
```

Einsetzen liefert

```python
xplus = z.subs('tau0', t0)
Math(r'x^+ = {}'.format(latex(xplus)))
```

also $x^+ = A^+ b$ mit

```python
Aplus = xplus.subs('b_1',1).subs('b_2',0).row_join(xplus.subs('b_1',0).subs('b_2',1))
Math(r'A^+ = {}'.format(platex(Aplus)))
```
