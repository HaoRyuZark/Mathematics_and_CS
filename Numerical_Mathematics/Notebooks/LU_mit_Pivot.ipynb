# Lösung

## Ohne Spalten-Pivot

Ist das Diagonalelement 0, dann wird mit der erstbesten Zeile mit einem Eintrag ungleich 0 getauscht

```python
def lu(AA, pivot = False):
    A = AA.copy()
    m,n = A.shape
    
    p = []
    for j in range(n-1):
        i0 = j
        if pivot:
            for i in range(j,m):
                if abs(A[i,j]) > abs(A[i0,j]): # Saplten-Pivot
                    i0 = i
        else:
            for i in range(j,m):
                if A[i,j] != 0: # kein Spalten-Pivot, nehme erstbeste Zeile
                    i0 = i
                    break
        
        p.append(i0+1)
        display(Math(r'Spalte\ ' + str(j+1) + r',\quad  p = ' + latex(p))) 

        A.row_swap(i0,j)
        out = latex(A)
        
        A[j+1:,j] /= A[j,j]
        out = out + r'\quad \Rightarrow \quad' + latex(A)
        
        A[j+1:,j+1:] -= A[j+1:,j] * A[j,j+1:]
        out = out + r'\quad \Rightarrow \quad' + latex(A)
        display(Math(out))

    L = eye(A.shape[0]) + np.tril(A, -1)
    U = Matrix(np.triu(A))

    return (L,U,p)

L, U, p = lu(A, pivot = False)
```

und somit

```python
Math( 'L=' + platex(L) + r',\quad U=' + platex(U) + r',\quad p=' + platex(p) )
```

## Mit Spalten-Pivot

```python
L, U, p = lu(A, pivot = True)
```

und somit

```python
Math( 'L=' + platex(L) + r',\quad U=' + platex(U) + r',\quad p=' + platex(p) )
```
