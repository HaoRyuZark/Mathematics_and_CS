# Aufgabenstellung

Bestimmen Sie mit Givens eine QR-Zerlegung von

```python
def Qg(n,i,j,cc,ss):
    no = sqrt(cc**2 + ss**2)
    c = cc/no
    s = ss/no
    Q = eye(n)
    Q[j,j], Q[j,i] =  c, s
    Q[i,j], Q[i,i] = -s, c    
    return Q

R = Matrix([[2,3,5],[0,7,11],[0,0,13]])
A = Qg(3,0,1,3,4).T*Qg(3,0,2,1,0).T*Qg(3,1,2,4,3).T*R*25
    
Math('A=' + platex(A))
```

# Lösung

```python
def givens(AA):
    A = AA.copy()
    n = A.shape[0]
    QQ = eye(n)
    for j in range(n-1):
        for i in range(j+1,n):
            no = sqrt(A[j,j]**2 + A[i,j]**2)
            c = A[j,j] / no
            s = A[i,j] / no
            Q = eye(n)
            Q[j,j], Q[j,i] =  c, s
            Q[i,j], Q[i,i] = -s, c
            out = r'Q_{{{0}{1}}} = {2},\qquad (QA)_{{{0}{1}}} = {2}{3} = {4}'.format(i+1,j+1,platex(Q),platex(A),platex(Q*A))
            display(Math(out))
            
            A = Q*A
            QQ = QQ*Q.T
    return QQ,A
```

```python
Q,R = givens(A)
```
