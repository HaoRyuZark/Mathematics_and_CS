# Lösung

## Interpolationspolynom nach Newton

Das Interpolationspolynom in *Newton-Darstellung* lautet

\begin{align*}
%\label{newtondarstellung}
p_n(x) =
d_{00}
+ d_{11} (x - x_0)
+ \ldots
+ d_{nn}(x-x_0) \cdot \ldots \cdot (x - x_{n-1}),
\end{align*}

mit *dividierten Differenzen*  $d_{ij}$

\begin{alignat*}{2}
d_{i0} &= y_i & i &= 0,\ldots,n \\
d_{ij} &= \frac{d_{ij-1} - d_{i-1j-1}}{x_i - x_{i-j}}  & \quad i &= 0,\ldots,n, \quad j=0,\ldots i.
\end{alignat*}


Für die dividierten Differenzen erhalten wir mit den gegebenen Werten

```python
def dd(xi,yi):
    n = len(xi)
    d = np.empty((n,n), dtype=object)
    
    d[:,0] = yi
    
    for k in range(1,n):
        for i in range(k,n):
            d[i,k] = (d[i,k-1] - d[i-1, k-1]) / (xi[i] - xi[i-k])
    
    return d

d = dd(ti,zi)

def dlatex(t, d):
    n = d.shape[0]
    dd = r'\begin{matrix} t_i & z_i \\\\'
    for i in range(n):
        dd += latex(t[i]) + r' & '
        for j in range(i+1):
            dd += platex(d[i,j]) + r' & '
        dd += r'\\'
    dd += r'\end{matrix}'
    return dd

Math(dlatex(ti,d))
```

und somit

```python
p = zeros(*zi[0].shape)
q = ones(*zi[0].shape)

t = symbols('t')

for k in range(len(ti)):
    p = p + matrix_multiply_elementwise(d[k,k], q)
    q = q * (t - ti[k])

Math('p(t) =' + latex(p) + '=' + latex(p.expand()))
```

```python
pp = lambdify(t, p, 'numpy')
tt = np.linspace(ti[0],ti[-1])
xy = np.array(list(map(pp, tt))).squeeze().T
plt.plot(xy[0], xy[1]);

xy = np.array(zi).T.squeeze()
plt.plot(xy[0], xy[1], 'ro');
```
