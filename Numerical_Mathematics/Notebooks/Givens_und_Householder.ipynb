# Lösung

## Givens

```python
def givens(AA):
    A = AA.copy()
    n = A.shape[0]
    QQ = eye(n)
    for j in range(n-1):
        for i in range(j+1,n):
            no = sqrt(A[j,j]**2 + A[i,j]**2)
            c = A[j,j] / no
            s = A[i,j] / no
            Q = eye(n)
            Q[j,j], Q[j,i] =  c, s
            Q[i,j], Q[i,i] = -s, c
            out = r'Q_{{{0}{1}}} = {2},\qquad (QA)_{{{0}{1}}} = {2}{3} = {4}'.format(i+1,j+1,platex(Q),platex(A),platex(Q*A))
            display(Math(out))
            
            A = Q*A
            QQ = QQ*Q.T
    return QQ,A
```

```python
Q,R = givens(A)
```

also

```python
Math(r'Q = {0}, \quad R = {1}, \quad QR = {2}'.format(*list(map(platex, (Q, R, Q*R)))))
```

## Householder

```python
def householder(AA):
    A = AA.copy()

    n = A.shape[0]

    d = zeros(n,1)

    Q = eye(A.shape[1]) 
    QA = A.copy()

    for j in range(n-1):
        print("**** Spalte {} ****".format(j+1))
        norma = A[j:,j].norm(2)

        d[j] = -sign(A[j,j]) * norma
        A[j,j] -= d[j]

        #display((A,d))

        normv2 = -2 * A[j,j] * d[j]

        Qk = eye(A.shape[1]) 
        Qk[j:, j:] = Qk[j:, j:] - 2/normv2 * A[j:,j] * A[j:,j].T
        QA = Qk * QA
        Q = Q * Qk
        for k in range(j+1, n):
            alpha = 2 * A[j:, j].dot(A[j:, k]) / normv2
            A[j:, k] -= alpha * A[j:, j] 

        display(Math(r"v = {} \qquad  Q_{} = {} \qquad \prod_k Q_k A = {}"
                     .format( platex(A[j:,j]), j+1, platex(Qk), platex(QA) ) ))
        display(Math(r"array = {} \qquad d = {}"
                     .format( latex(A), latex(d) ) ))

    print("**** Spalte {} ****".format(n))
    d[-1] = A[-1,-1]
    A[-1,-1] = nan

    display(Math(r"array = {} \qquad d = {}"
                     .format( latex(A), latex(d) ) ))

    R = Matrix(triu(A,1) + diag(flatten(d), unpack = True))
    
    return Q,R
```

```python
Q,R = householder(A)
```

```python
Math(r'Q = {0}, \quad R = {1}, \quad QR = {2}'.format(*list(map(platex, (Q, R, Q*R)))))
```
