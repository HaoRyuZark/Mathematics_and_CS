# Aufgabenstellung

Prager Oettli für

```python
A = Matrix([[2,-1],[1,1]])
b = Matrix([1, 2])
xs = Matrix([0.9, 1.1])

B = ones(2,2)/10
c = ones(2,1)/10

Math(r'A = {}, \quad b = {}, \quad |\Delta A| \leq {}, \quad |\Delta b| \leq {}, \quad \tilde{{x}} = {}'.format(platex(A), platex(b), platex(B), platex(c), platex(xs)))
```

# Lösung

Für $x \in \mathbb{R}^n$, $A \in \mathbb{R}^{n \times n}$ sei
$$
|x| = (|x_1|, \ldots, |x_n| )^T,
\qquad
|A| = \bigl(|a_{ij}|\bigr)_{i,j}.
$$
Ist $\det(A) \ne 0$ und $\tilde{x}$ eine Näherungslösung von $Ax = b$ mit Residuum
$\tilde{r} = b - A \tilde{x}$ sowie
$$
B \in \mathbb{R}^{n \times n},\ c \in \mathbb{R}^n,
\qquad
b_{ij} \ge 0,\quad c_i \ge 0 \qquad\forall i,j
$$
mit
$$
| \tilde{r} | \le B | \tilde{x} | + c.
$$
Dann gibt es $| \Delta A | \le B$ und $| \Delta b | \le c$ so dass
$\tilde{x}$ exakte Lösung von
$$
(A + \Delta A) \tilde{x} = b + \Delta b
$$
ist. Die Ungleichheitszeichen sind jeweils komponentenweise anzuwenden.

```python
Math(r'B = {}, \quad c = {}'.format(platex(B), platex(c)))
```

```python
rs = b - A*xs
Math(r'\tilde{{r}} = {}'.format(platex(rs)))
```

```python
s = B*abs(xs) + c
Math(r's = {}'.format(platex(s)))
```

Vergleich

```python
[rsk <= sk for rsk,sk in zip(rs, s)]
```
