# Relaxierter Jacobi

Bestimmen Sie den optimalen Parameter $\omega$ für das relaxierte Jacobi-Verfahren
zur Systemmatrix

```python
A = Matrix([[2, -1, 0], [-1, 4, -1], [-1, -1, 4]])
Math('A=' + platex(A))
```

## Lösung

Hat $B_J = D^{-1}(E+F)$ reelle Eigenwerte $-1 < \lambda_1 \leq \ldots \leq \lambda_n < 1$,
dann ist der optimale Parameter gegeben durch
$$
\omega_\text{opt} = \frac{2}{2-\lambda_1-\lambda_n}.
$$
Hier ist

```python
from numpy import triu, tril, diag

D = Matrix(diag(diag(A)))
E = - Matrix(tril(A, -1))
F = - Matrix(triu(A,  1))

B = D.inv() * (E + F)

Math(r'B_J = D^{-1}(E+F) =' + platex(D.inv()) + platex(E+F) + '=' + platex(B))
```

mit Eigenwerten

```python
ev = list(B.eigenvals().keys())
ev.sort()

out = ''
for i,e in enumerate(ev):
    out += r'\lambda_' + str(i+1) + '=' + latex(e) + r', \quad'

Math(out)
```

und somit ist

```python
om = 2 / (2- max(ev) - min(ev))
Math(r'\omega_\text{{opt}}= {}.'.format(latex(om)))
```
