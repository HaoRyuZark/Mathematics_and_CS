# Lösung

## Gerschgorin

Nach Gerschgorin gilt: alle Eigenwerte $\lambda_i$ von $A$ liegen in
$$
K = \bigcup_{i = 1}^n K_i, \qquad K_i = \{ z\ |\ z \in \mathbb{C},\
|z - a_{ii}| \le \sum_{j \not= i} |a_{ij}| \}
$$
der Vereinigung der Kreisscheiben $K_i$.

Für die Mittelpunkte und Radien der Kreisscheiben erhalten wir

```python
m,n = A.shape

def plotcircle(m, r, nphi=36):
    phi = pl.linspace(0.0, 2.0*pl.pi, nphi)
    pl.plot(m + r * pl.cos(phi), r * pl.sin(phi))

for i in range(m):
    mi = A[i,i]
    ri = sum(A[i,:].applyfunc(abs)) - abs(mi)
    display(Math(r'K_' + str(i+1) + r':\quad m = ' + str(mi)+ r',\quad r = ' + str(ri) ) )
    plotcircle(mi, ri)

dummy = pl.axis('equal')
```

## Eigenwerte

```python
ev = list(A.eigenvals().keys())

out = ''
for i,e in enumerate(ev):
    out += r'\lambda_' + str(i+1) + '=' + latex(e) + r', \quad'

Math(out)
```

## Vektoriteration

Die Vektoriteration zum Startwert
 $x^{(0)}$ mit $\| x^{(0)} \|_2 = 1$
ist gegeben durch:
\begin{align*}
y^{(k+1)} &= Ax^{(k)}\\
x^{(k+1)} &= \frac{y^{(k+1)}}{\|y^{(k+1)}\|_2}\\
\mu^{(k+1)} &= < x^{(k)}, y^{(k+1)}>
\end{align*}

Wir erhalten in unserem Fall

```python
x0 = Matrix([0,1,0])

nit = 3
st  = 6
x = x0.copy()
for k in range(nit):
    y   = A * x
    mue = x.T * y
    x   = y / y.norm(2)
    
    display(Math(r'\mu^{(' + str(k+1) + r')} = ' + latex(mue[0]) + r'\approx' + latex(mue[0].evalf(st))
                 + r', \qquad x^{(' + str(k+1) + r')}\approx' + platex(x.evalf(st)) ) )
```
