```python
from sympy import *

from IPython.display import display, Math, Latex
from sympy.interactive import printing
printing.init_printing(use_latex='mathjax')
platex = lambda A: latex(A,mat_str='pmatrix',mat_delim='')

import pylab as pl
%matplotlib inline
```

# Aufgabenstellung

Bestimmen Sie mit Hilfe von Givens-Rotationen die Hessenberg-Transformation der
Matrix

```python
A = Matrix([[2,3,4],[3,-50,25],[4,25,-75]])
Math('A=' + platex(A))
```

## Lösung

Wir betrachten die verkürzte erste Spalte

```python
u = A[1:,0]
Math('u=' + platex(u))
```

und bestimmen die zugehörige Givens-Matrix $\tilde{Q}$, die u auf ein Vielfaches des ersten Einheitsvektors transformiert

```python
c = u[0] / u.norm(2)
s = u[1] / u.norm(2)

Qt = Matrix([[c,s],[-s,c]])

Math(r'\tilde{Q}=' + platex(Qt) + r',\qquad \tilde{Q} u =' + platex(Qt) + platex(u) + '=' +platex(Qt*u))
```

Mit $\tilde{Q}$ erzeugen wir $Q$ als

```python
Q = eye(A.shape[0])
Q[1:,1:] = Qt
Math('Q=' + platex(Q))
```

und mit $H = Q A Q^T$ erhalten wir die Hessenberg-Transformierte

```python
H = Q*A*Q.T
Math('H=' + platex(H))
```
