# Lösung

## LU-Zerlegung

```python
L, U, p = A.LUdecomposition()
P=eye(A.shape[0]).permuteFwd(p)
Math(r'L = ' + platex(L) + r', \quad U = '+ platex(U)+ r', \quad P = '+ platex(P))
```

Die exakte Lösung lautet

```python
x = A.solve(b)
Math('x = ' + platex(x))
```

## Nachiteration

### Startwert $\tilde{L}\tilde{U}x_0 = b$

Durch vorwärts Einsetzen lösen wir zunächst $\tilde{L} y_0 = b$

```python
y0 = Lt.solve(b)
Math(r'y_0 =' + platex(y0))
```

und durch rückwärts Einsetzen $\tilde{U} x_0 = \tilde{y}$

```python
x0 = Ut.solve(y0)
Math(r'x_0 =' + platex(x0))
```

### Nachiteration

```python
def nachit(x0, A, Lt, Ut, b):
    x = x0.copy()
    r = b - A*x
    z = Lt.solve(r)
    p = Ut.solve(z)
    x += p
    
    out = r'r = {0},\quad z = {1}, \quad p = {2}, \quad x = {3}'.format(*list(map(platex,[r,z,p,x])))
    display(Math(out))

    return x

x1 = nachit(x0, A, Lt, Ut, b)
```

```python
e0 = x-x0
Math(r"e_0 = x - x_0 = " + platex(e0) + ",\qquad \|e_0\|_\infty = {}".format(latex(e0.norm(oo))))
```

```python
e1 = x-x1
Math(r"e_1 = x - x_1 = " + platex(e1) + ",\qquad \|e_1\|_\infty = {}".format(latex(e1.norm(oo))))
```
