# Lösung

## Mindestens eine Nullstelle

$f$ ist auf dem Intervall stetig und es gilt

```python
display(Math('f(a)=' + latex(f(a)) + r'\approx' + latex(f(a).evalf(7))))
display(Math('f(b)=' + latex(f(b)) + r'\approx' + latex(f(b).evalf(7))))
```

d.h. es muss mindestens ein Vorzeichenwechsel stattfinden.

## Bisektion

```python
l = Rational(1,4)

def bisektion(f, a, b, l):
    while (b-a) > l:
        x = (a + b)/Rational(2)
        
        s = r'a = {0}, \quad x = {1}, \quad b = {2}, \qquad f(a) = {3}, \quad f(x) = {4}, \quad f(b) = {5}'
        axb  = [a, x, b]
        faxb = [f(z).evalf(6) for z in axb]
        display(Math(s.format(*(axb + faxb))))
        
        if f(a)*f(x) > 0:
            a = x
        else:
            b = x
    
    return (a,b)

an, bn = bisektion(f, a, b, l)
                
Math(r'[a_{\text{neu}}, b_{\text{neu}}] = ' + latex([an, bn]))
```

## Newton

```python
def newton(f, x0, nit = 3):
    x = symbols('x')
    fs = Lambda(x, f(x).diff(x))
    
    x = x0

    display(Math(r'x_0 =' + latex(x.evalf(6))))
    
    for i in range(nit):
        x = x - f(x) / fs(x)
        display(Math(r'x_' + str(i+1) + r'\approx' + latex(x.evalf(6))))
    
    return x
        
x = newton(f, (an + bn) / 2.0)
```
