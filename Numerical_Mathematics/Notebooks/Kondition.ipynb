```python
from sympy import *

from IPython.display import display, Math, Latex
from sympy.interactive import printing
printing.init_printing(use_latex='mathjax')
platex = lambda A: latex(A,mat_str='pmatrix',mat_delim='')

import pylab as pl
%matplotlib inline
```

# Aufgabenstellung

Absolute Konditionszahl von $f$ an $x_0 = 1$ für $|y-x_0|\leq \delta$ und $\delta = 2$

* exakt

* obere Schranke mit Mittelwertsatz

* mit differentieller Fehleranalyse

für

```python
x = symbols('x')

f = Lambda(x, x**3 -4*x**2)

x0 = 1
delta = 2

a = x0 - delta
b = x0 + delta

display(Math('f(x)=' + latex(f(x))))

dummy = plot(f(x), (x,a,b))
```

# Lösung

## Exakt

Die absolute Konditionszahl $\kappa_a$ an $x_0$ zu $\delta$ ist die kleinste Konstante s.d.
$$
|f(y) - f(x_0)| \leq \kappa_a |y - x_0| \qquad \forall y \quad\text{mit}\quad |y-x_0|\leq\delta
$$
Damit gilt also
$$
\kappa_a = \max_{y, \ |y-x_0|\leq\delta} \Big| \frac{f(y) - f(x_0)}{y - x_0} \Big|.
$$
Wir suchen deshalb das Betragsmaximum von 
$$
q(y) = \frac{f(y) - f(x_0)}{y - x_0}
$$
bezüglich $y$ über dem Bereich $|y-x_0|\leq\delta$.

```python
y = symbols('y')

q = Lambda(y, ((f(y)-f(x0))/(y-x0)).simplify())

display(Math('q(y) = ' + latex(q(y))))
dummy = plot(q(y),(y,a,b))
```

Extremstellen bestimmen

```python
display(Math("q'(y) = " + latex(q(y).diff(y))))
xex = solve(q(y).diff(y),y)
display(*xex)
```

Funktionswerte von $q$ an den Extremstellen und den Randpunkten

```python
xex += [a,b]
qex = list(map(q, xex))
#display(xex)
display(*qex)
```

Betragsmaximum davon liefert die Konditionszahl

```python
kappa = max(list(map(abs, qex)))
display(kappa)
```

## Mittelwertsatz

Der Mittelwertsatz liefert die Abschätzung
$$
|f(y) - f(x_0)| \leq \max_{\xi\in[x_0-\delta,x_0+\delta]}|f'(\xi)| \: |y - x_0| 
\qquad \forall y \quad\text{mit}\quad |y-x_0|\leq\delta
$$
d.h. als obere Schranke für $\kappa_a$ kann
$$
s = \max_{\xi\in[x_0-\delta,x_0+\delta]}|f'(\xi)|
$$
benutzt werden. Dazu müssen wir $f'(x)$ betragsmäßig maximieren.

```python
fs = Lambda(x, f(x).diff(x).simplify())

display(Math("f'(x) = " + latex(fs(x))))
dummy = plot(fs(x),(x,a,b))
```

Extremstellen bestimmen

```python
display(Math("f''(x) = " + latex(fs(x).diff(x))))
xex = solve(fs(x).diff(x),x)
display(*xex)
```

Funktionswerte von $f'$ an den Extremstellen und den Randpunkten

```python
xex += [a,b]
fsex = list(map(fs, xex))
#display(xex)
display(*fsex)
```

Betragsmaximum davon liefert die obere Schranke für die Konditionszahl

```python
s = max(list(map(abs, fsex)))
display(s)
```
