# Grundlagen

$Q \in \mathbb{R}^{n \times n}$ heißt *orthonormal*, falls
die Spaltenvektoren $q_j$ paarweise senkrecht stehen und Länge $1$ haben, d.h.
$$
q_i^T q_j = \delta_{ij} =
\begin{cases}
1 & i = j\\ 0 & \text{sonst}
\end{cases}.
$$

Sind $Q, \tilde{Q} \in \mathbb{R}^{n \times n}$ orthonormal, dann gilt:

-  $QQ^T = Q^TQ = I$, d.h. $Q^{-1} = Q^T$
- $\kappa_2(Q) = 1$
- $Q\tilde{Q}$ ist orthonormal

# Beispiele

## Drehungen

Die Matrix
$$
Q = \begin{pmatrix}
 c & s\\
-s & c
\end{pmatrix},
\qquad
c = \cos(\alpha),
\quad
s = \sin(\alpha)
$$
definiert eine Drehung in der $x$-$y$-Ebene im Uhrzeigersinn.

Wir betrachten die Beispiele

```python
a = symbols('a')
Qd = Lambda(a, Matrix([[cos(a), sin(a)], [-sin(a), cos(a)]]))

a1 = 30
Q1 = Qd(a1 * pi/180)
Math(r'\alpha = {}^\circ \qquad  Q_1 = '.format(a1) + platex(Q1))
```

```python
def Qtest(Q, nv = 3):
    x = pl.zeros((2,2))

    pl.seed(17)
    for k in range(nv):
        x[:,1] = pl.rand(2)

        pl.plot(*x, 'b', label = r'$x_{}$'.format(k+1))
        pl.quiver(*x[:,0], *x[:,1], scale_units='xy', scale=1, color = (0,0,1))

        Qx = pl.array(Q * x).astype(pl.float64)

        pl.plot(*Qx, 'r', label = r'$Qx_{}$'.format(k+1))
        pl.quiver(*Qx[:,0], *Qx[:,1], scale_units='xy', scale=1, color = (1,0,0))

    pl.axis('equal')
    pl.legend()

Qtest(Q1)
```

oder

```python
a2 = 60
Q2 = Qd(a2 * pi/180)
Math(r'\alpha = {}^\circ \qquad  Q_2 = '.format(a2) + platex(Q2))
```

```python
Qtest(Q2)
```

Als Produkt erhalten wir

```python
Math(r'Q_1 Q_2 = '+ platex(Q1) + platex(Q2) + "=" + platex(Q1*Q2))
```

```python
Math(r'Q_2 Q_1 = '+ platex(Q2) + platex(Q1) + "=" + platex(Q2*Q1))
```

Wie erwartet stimmt das überein mit einer Drehung um die Summe der beiden Winkel

```python
a3 = a1 + a2
Q3 = Qd(a3 * pi/180)
Math(r'\alpha = {}^\circ \qquad  Q_3 = '.format(a3) + platex(Q3))
```

## Spiegelungen

- $Q = I - 2ww^T,\ \|w\|_2 = 1$ definiert eine Spiegelung an der Hyperebene $w^Tx = 0$,
d.h. an der Hyperebene die senkrecht auf $w$ steht
- $Q = Q^T = Q^{-1}$ und damit orthonormal
- $\det(Q) = -1$ d.h. die Orientierung kehrt sich um

Betrachten wie als Beispiel die Hyperebene $H$, die
senkrecht auf

```python
w = Matrix([3,4])
w = w / w.norm(2)
Math(r'w = ' + platex(w))
```

steht. Für die Spiegelmatrix erhalten wir

```python
Q = eye(2) - 2 * w * w.T
Math('Q = I - 2ww^T = {} - 2 {} = {}'.format(platex(eye(2)), platex(w * w.T), platex(Q)))
```

```python
def Qtest2(Q, w, nv = 3):
    ww = pl.zeros((2,2))
    ww[:,1] = pl.array(w).astype(pl.float64).flatten()
    pl.plot(*ww, color = (0,1,0), label = r'$w$')
    pl.quiver(*ww[:,0], *ww[:,1], scale_units='xy', scale=1, color = (0,1,0))

    h  = Matrix([w[1], -w[0]])
    hh = pl.zeros((2,2))
    hh[:,1] = pl.array(h).astype(pl.float64).flatten()
    pl.plot(*hh, color = (0,1,0), label = "H")
    pl.plot(*(-hh), color = (0,1,0))

    Qtest(Q)

Qtest2(Q, w)
```

Für die Determinante erhalten wir

```python
Math(r'\det(Q) =' + latex(Q.det()))
```

Für

```python
w = Matrix([-1, 2])
w = w / w.norm(2)
Math(r'w = ' + platex(w))
```

ergibt sich analog

```python
Q = eye(2) - 2 * w * w.T
Math('Q = I - 2ww^T = {} - 2 {} = {}'.format(platex(eye(2)), platex(w * w.T), platex(Q)))
```

und

```python
Qtest2(Q, w)
```

## Spiegelmatrizen im $\mathbb{R}^2$

In $\mathbb{R}^2$ lassen sich alle Normalenvektoren $w$ in der Form

```python
a,b = symbols('alpha,beta')

w = Matrix([cos(b), sin(b)])

Math('w = ' + platex(w))
```

schreiben. Damit ergibt sich für $Q$

```python
Q = eye(2) - 2 * w * w.T
Math('Q = I - 2ww^T = {} - 2 {} = {}'.format(platex(eye(2)), platex(w * w.T), platex(Q)))
```

Benutzt man nun die Additionstheoreme, so vereinfacht sich das zu

```python
Q.simplify()
Math('Q = {}'.format(platex(Q)))
```

Mit $\alpha = -2\beta$ folgt schließlich

```python
Math('Q = {}'.format(platex(Q.subs(b, -a/2))))
```
