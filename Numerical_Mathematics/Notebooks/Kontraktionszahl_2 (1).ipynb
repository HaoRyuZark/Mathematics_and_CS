# Lösung

##  $\Phi(X)\subset X$

Für $\Phi$ erhalten wir wegen

```python
fs = Lambda(x, f(x).diff(x).simplify())
display(Math(r"f'(x) = " + latex(fs(x))))

Phi   = Lambda(x, (x - f(x)/fs(x)).simplify())
display(Math(r"\Phi(x) = " + latex(Phi(x))))
```

und somit

```python
Phis  = Lambda(x, Phi(x).diff(x).simplify())
Phiss = Lambda(x, Phis(x).diff(x).simplify())

display(Math(r"\Phi'(x) = " + latex(Phis(x))))
display(Math(r"\Phi''(x) = " + latex(Phiss(x))))
```

Wir bestimmen für $\Phi$ die Funktionswerte an allen Stellen $x$ im Intervall $[a,b]$ mit $\Phi'(x)=0$ sowie die Werte am Rand des Intervalls.
Minimum und Maximum der Funktionswerte von $\Phi$ sind dann jeweils Minimum bzw. Maximum dieser Werte.

```python
def minmax(f, a, b):
    x = symbols('x')
    fs = Lambda(x, f(x).diff(x))
    
    xe = solve(fs(x),x)
    xe = [x for x in xe if (a<=x) and (x<=b)]
    fe = list(map(f, xe)) 
    
    xr = [a,b]
    fr = list(map(f, xr))
    
    return(xe, xr, fe, fr)
```

Wir erhalten für $\Phi$ die Werte

```python
xe, xr, Phie, Phir = minmax(Phi, a, b)
Phier = Phie + Phir

Phie, Phir
```

und damit

```python
display(Math(r'\min_{x\in[a,b]} \Phi(x) = ' + latex(min(Phier))))
display(Math(r'\max_{x\in[a,b]} \Phi(x) = ' + latex(max(Phier))))
```

## Kontraktionszahl $\alpha$

Nach dem Mittelwertsatz gilt

\begin{align*}
|\Phi(y) - \Phi(x)| = |\Phi'(\xi)|\ |y-x|, \quad \xi\in\text{co}(x,y)
\end{align*}

und somit

\begin{align*}
\alpha \leq \max_{\xi \in X} |\Phi'(\xi)|.
\end{align*}

Wenn wir also das Betragsmaximum von $\Phi'$ über $X$ bestimmen, erhalten
wir (eine obere Schranke für) $\alpha$.

Um das Betragsmaximum zu bestimmen, berechnen wir das Minimum bzw. Maximum von $\Phi'$ auf $X$.
Dazu betrachten wir wieder die Extremwerte und die Werte am Rand

```python
display(Math(r"\Phi'(x) = " + latex(Phis(x))))
display(Math(r"\Phi''(x) = " + latex(Phiss(x))))
```

Wir erhalten für $\Phi'$ die Werte

```python
xe, xr, Phise, Phisr = minmax(Phis, a, b)
Phiser = Phise + Phisr

Phise, Phisr
```

und damit

```python
display(Math(r"\min_{x\in[a,b]} \Phi'(x) = " + latex(min(Phiser))))
display(Math(r"\max_{x\in[a,b]} \Phi'(x) = " + latex(max(Phiser))))
```

und damit

```python
display(Math(r"\max_{x\in[a,b]} |\Phi'(x)| = " + latex(max(map(abs, Phiser)))))
```
