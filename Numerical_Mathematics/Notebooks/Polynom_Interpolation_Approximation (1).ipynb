# Lösung

## Interpolationspolynom nach Newton

Das Interpolationspolynom in *Newton-Darstellung* lautet

\begin{align*}
%\label{newtondarstellung}
p_n(x) =
d_{00}
+ d_{11} (x - x_0)
+ \ldots
+ d_{nn}(x-x_0) \cdot \ldots \cdot (x - x_{n-1}),
\end{align*}

mit *dividierten Differenzen*  $d_{ij}$

\begin{alignat*}{2}
d_{i0} &= y_i & i &= 0,\ldots,n \\
d_{ij} &= \frac{d_{ij-1} - d_{i-1j-1}}{x_i - x_{i-j}}  & \quad i &= 0,\ldots,n, \quad j=0,\ldots i.
\end{alignat*}


Für die dividierten Differenzen erhalten wir mit den gegebenen Werten

```python
def dd(xi,yi):
    n = len(xi)
    d = zeros(n)
    
    d[:,0] = yi
    
    for k in range(1,n):
        for i in range(k,n):
            d[i,k] = (d[i,k-1] - d[i-1, k-1]) / (xi[i] - xi[i-k])
    
    return d

d = dd(xi,yi)

display(d)
```

und somit

```python
p = 0
q = 1

x = symbols('x')

for k in range(len(xi)):
    p += d[k,k] * q
    q *= (x - xi[k])

Math('p(x) =' + latex(p) + '=' + latex(p.expand()))
```

## Ausgleichspolynom

```python
t = symbols('t')
p = Poly(pa.subs(x,t), t)
Math('p(x) =' + latex(p(x)))
```

Für das Ausgleichspolynom muss 
$\big\| Av-y \big\|_2^2$
mit

```python
n = len(xi)

m = p.degree()
cs = p.all_coeffs()
cs.reverse()

A = Matrix(n, 0, [])
v = []

x = symbols('x')
xpot = []

k = 0
for c in cs:
    if c != 0:
        xii = Matrix([w**k for w in xi])
        A = A.row_join(xii)
        v.append(c)
        xpot.append(x**k)
    k += 1

v = Matrix(v)
xpot = Matrix(xpot)

display(Math('A=' + platex(A) + r',\qquad v=' + platex(v) + r',\qquad y=' + platex(Matrix(yi))))
```

in $v$ minimiert werden. Da $A$ vollen Rang hat, ist der Minimierer die eindeutige Lösung der Normalgleichung

\begin{align*}
A^TA v= A^Ty.
\end{align*}
Mit

```python
ATA = A.T*A
ATy = A.T*Matrix(yi)

display(Math(r'A^T A=' + platex(ATA) + r', \qquad A^T y=' + platex(ATy)))
```

folgt

```python
xx = ATA.solve(ATy)

display(Math('v = ' + platex(xx)))
```

und somit

```python
pp = (xx.T * xpot)[0]

display(Math('p(x)=' + latex(pp)))
```
